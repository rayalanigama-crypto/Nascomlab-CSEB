```python
import sympy as sp
x = sp.symbols('x')
F = 40 + 5*x - 0.2*x**2
F_prime = sp.diff(F, x)
print("F'(x):", F_prime)
solution = sp.solve(F_prime, x)
print("Value of x when F'(x) = 0:", solution)
print("The predicted score increases when study hours are less than 12.5.")
print("At 12.5 hours, the derivative is zero.")
print("Beyond 12.5 hours, the predicted score decreases according to the model.")
```
